Importing neccessary Libraries

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import numpy as np
```

Exactracting and reading the Data

```python
df = pd.read_csv('https://raw.githubusercontent.com/krishnaik06/playstore-Dataset/main/googleplaystore.csv')
```

The dataset has inconsistent values and missing values

```python
df[~df['Reviews'].str.isnumeric()]
```

In Reviews feature one record with '10472' has 3.0M which will be further difficult to typcast and FE

```python
df_copy = df.copy()
```

```python
df_copy = df_copy.drop(df_copy.index[10472])
```

```python

df_copy[~df_copy['Reviews'].str.isnumeric()]
```

Converting Reviews Feature to int type (typecasting)

```python
df_copy['Reviews'] = df_copy['Reviews'].astype(int)
```

```python
df_copy['Reviews'].dtype
```

```python
df_copy['Size'].unique()
```

```python
df_copy['Size'] = df_copy['Size'].str.replace('M','000')
df_copy['Size'] = df_copy['Size'].str.replace('k','')
df_copy['Size'] = df_copy['Size'].replace('Varies with device',np.nan)
df_copy['Size'] = df_copy['Size'].astype(float)
```

```python
df_copy['Size'].unique()
```

```python
df_copy['Installs'].unique()
```

```python
df_copy['Price'].unique()
```

Columns Price and Installs have special chars

```python
char_to_remove = ['+',',','$']
col_to_clean = ['Installs', 'Price']
```

```python
for item in char_to_remove:
    for col in col_to_clean:
        df_copy[col] = df_copy[col].str.replace(item,'')
```

```python
df_copy['Price'].unique()
```

```python
df_copy['Installs'].unique()
```

```python
df_copy['Price'] = df_copy['Price'].astype(float)
df_copy['Installs'] = df_copy['Installs'].astype(int)
```

```python
df_copy.info()
```

```python
df_copy['Last Updated'].unique()
```

feature Lastupdated has data in Jan,12,2222. Convert this to datetime

```python
df_copy['Last Updated'] = pd.to_datetime(df_copy['Last Updated'])
```

```python
df_copy['Day'] = df_copy['Last Updated'].dt.day
df_copy['Month'] = df_copy['Last Updated'].dt.month
df_copy['Year'] = df_copy['Last Updated'].dt.year
```

```python
df_copy.info()
```

```python
df_copy.to_csv('cleaned_data.csv')
```

There were some records with duplicates. Need to drop them

```python
df_copy = df_copy.drop_duplicates(subset=['App'], keep='first')
```

```python
df_copy.duplicated('App').sum()
```

```python
numerical_features = [feature for feature in df_copy.columns if df_copy[feature].dtype != 'O']
categorical_features = [feature for feature in df_copy.columns if df_copy[feature].dtype == 'O']

print('There are {} Numerical features: {}'.format(len(numerical_features), numerical_features))
print('There are {} Categorical features: {}'.format(len(categorical_features), categorical_features))
```

Count of Categorical features

```python
for col in categorical_features:
    print(df_copy[col].value_counts(normalize=True)*100)
    print('------------------------------')
```

Count data of numercial features

```python
plt.figure(figsize=(15,15))
plt.suptitle("Analysis of Numerical Features", fontsize=20, fontweight='bold', alpha=0.8, y=1.)

for i in range(0, len(numerical_features)):
    plt.subplot(5,3,i+1)
    sns.kdeplot(x=df_copy[numerical_features[i]], shade=True, color='r')
    plt.xlabel(numerical_features[i])
    plt.tight_layout()
```

Rating and Year is left skewed
Reviews, Installs, Size, Price were right skewed

```python
plt.figure(figsize=(20,15))
plt.suptitle("Analysis for categorical Data", fontsize=20, fontweight='bold', alpha=0.8,y=1.)
category = ['Type', 'Content Rating']

for i in range(0, len(category)):
    plt.subplot(2,2,i+1)
    sns.countplot(x=df_copy[category[i]], palette='Set2')
    plt.xlabel(category[i])
    plt.xticks(rotation=45)
    plt.tight_layout()
```
